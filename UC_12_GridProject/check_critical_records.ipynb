```python
# Read the Gold incident notes table
df = spark.table(
    "uc_grid_sensor.grid_gold.gold_incident_notes"
)

# Count critical records
critical_count = (
    df.filter("upper(severity) = 'CRITICAL'")
      .count()
)

# Check whether any critical records exist
has_critical = critical_count > 0

# Pass the result to the Databricks Job
dbutils.jobs.taskValues.set(
    key="has_critical",
    value=has_critical
)

print(f"Critical records found: {critical_count}")
print(f"Should route to Lakebase: {has_critical}")
```
